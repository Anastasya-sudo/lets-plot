## geom_tile — small cells: solid fill, no jitter
Here the fill **color is the data** - each cell's value is read by its color, so a solid fill keeps the heatmap clean and readable. With many small cells the edges are too short to show 
hand-drawn jitter (and scribble fill would only turn the data into noise). The xkcd look still comes through via the wobbly **axes**.

```python
import numpy as np
np.random.seed(3)
days = list(range(1, 31))
hours = list(range(24))
gd, gh = np.meshgrid(days, hours)
activity = (np.sin(gh / 3.0) + 1) * 50 + np.random.normal(0, 10, gd.size).reshape(gh.shape)

ggplot(
      {
          'day': gd.flatten(),
          'hour': gh.flatten(),
          'activity': activity.flatten()
      },
      aes('day', 'hour', fill='activity')
  ) + geom_tile() + \
      labs(x='Day of month', y='Hour') + \
      ggtitle('MY SCREEN TIME, HOUR BY HOUR') + \
      base
```

## geom_hex

```python
import numpy as np
np.random.seed(42)
n = 2000

ggplot(
      {
          'x': np.random.normal(0, 1, n),
          'y': np.random.normal(0, 1, n)
      },
      aes('x', 'y')
  ) + geom_hex(bins=[4, 4], color='white', size=1.0) + \
      labs(x='x', y='y') + \
      ggtitle('JUST A FEW BIG HEXAGONS') + \
      base
```

## geom_hex — small cells: solid fill, no jitter
Here the fill **color is the data** — each cell's value is read by its color, so a solid fill keeps the heatmap clean and readable. With many small cells the edges are too short to show 
hand-drawn jitter (and scribble fill would only turn the data into noise). The xkcd look still comes through via the wobbly **axes**.

```python
np.random.seed(42)
n = 10000
  
ggplot(
      {
          'x': np.random.normal(0, 1, n),
          'y': np.random.normal(0, 1, n)
      },
      aes('x', 'y')
  ) + geom_hex(bins=[30, 30]) + \
      labs(x='x', y='y') + \
      ggtitle('DENSITY OF 10,000 POINTS (HEXBIN)') + \
      base
```

## variadic lines

```python
ggplot(
    {
        'day':  list(range(1, 11)) * 2,
        'temp': [62, 64, 67, 70, 68, 72, 75, 73, 78, 80,
                 55, 58, 57, 60, 63, 61, 66, 64, 69, 72],
        'wind': [12.0, 10.5, 8.0, 6.5, 9.0, 5.0, 4.0, 6.0, 3.5, 2.0,
                 14.0, 13.0, 11.5, 9.5, 8.0, 10.0, 7.0, 8.5, 5.5, 4.5],
        'month': ['May'] * 10 + ['June'] * 10
    },
    aes('day', 'temp', size='wind', color='wind', group='month')
) + geom_line() + \
    scale_size([0.5, 6.0]) + \
    scale_color_hue(h=[0, 150], h_start=240) + \
    labs(x='Day', y='Temp', size='Wind', color='Wind') + \
    ggtitle('THE WIND THAT WOULD NOT COMMIT') + \
    ggsize(700, 500) + \
    base
```
